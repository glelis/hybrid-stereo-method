```python
from PIL import Image
import numpy as np

def media_imagens(imagens):
    # Verifica se a lista contém 12 imagens
    if len(imagens) != 12:
        raise ValueError("A lista deve conter exatamente 12 imagens.")
    
    # Abre as imagens e converte para arrays numpy
    arrays = [np.array(imagem) for imagem in imagens]
    
    # Calcula a média ao longo do eixo 0
    media_array = np.mean(arrays, axis=0).astype(np.uint8)
    
    # Converte o array de volta para uma imagem
    imagem_media = Image.fromarray(media_array)
    
    return imagem_media
```

```python
import os
import datetime

def explore_directory(root_path, max_depth=None):
    """
    Explore a directory, all archives and subdirectories, and return a comprehensive list
    
    Args:
        root_path (str): Path to the directory to explore
        max_depth (int, optional): Maximum depth to explore (None for unlimited)
        
    Returns:
        dict: Information about the directory structure, files and stats
    """
    result = {
        'root': root_path,
        'directories': [],
        'files': [],
        'stats': {
            'total_dirs': 0,
            'total_files': 0,
            'total_size': 0
        },
        'by_extension': {}
    }
    
    for dirpath, dirnames, filenames in os.walk(root_path):
        # Calculate current depth
        rel_path = os.path.relpath(dirpath, root_path)
        current_depth = 0 if rel_path == '.' else rel_path.count(os.sep) + 1
        
        # Stop if we've reached maximum depth
        if max_depth is not None and current_depth > max_depth:
            dirnames.clear()  # Don't explore deeper
            continue
            
        # Add directory info
        dir_info = {
            'path': dirpath,
            'name': os.path.basename(dirpath),
            'files_count': len(filenames),
            'subdirs_count': len(dirnames),
            'depth': current_depth
        }
        result['directories'].append(dir_info)
        result['stats']['total_dirs'] += 1
        
        # Add file info
        for filename in filenames:
            file_path = os.path.join(dirpath, filename)
            try:
                file_size = os.path.getsize(file_path)
                extension = os.path.splitext(filename)[1].lower()
                
                file_info = {
                    'name': filename,
                    'path': file_path,
                    'size': file_size,
                    'extension': extension,
                    'modified': datetime.datetime.fromtimestamp(os.path.getmtime(file_path)).strftime('%Y-%m-%d %H:%M:%S')
                }
                
                result['files'].append(file_info)
                result['stats']['total_files'] += 1
                result['stats']['total_size'] += file_size
                
                # Track file extensions
                if extension not in result['by_extension']:
                    result['by_extension'][extension] = {'count': 0, 'size': 0}
                result['by_extension'][extension]['count'] += 1
                result['by_extension'][extension]['size'] += file_size
                
            except Exception as e:
                print(f"Error accessing {file_path}: {e}")
    
    return result

# Example usage:
directory_structure = explore_directory('/home/lelis/Documents/Projetos/hybrid-stereo-method/data/raw/hybrid_stereo/2025-03-08-stQ-melon24-amb0.00-glo0.50')
```

```python
def find_files_by_name(directory_structure, target_name, partial_match=False):
    """
    Find files with a specific name in the directory structure.
    
    Args:
        directory_structure (dict): The directory structure from explore_directory
        target_name (str): The target filename to search for
        partial_match (bool): If True, find files that contain target_name as a substring
        
    Returns:
        list: List of file paths matching the target name
    """
    matching_paths = []
    
    for file_info in directory_structure['files']:
        if (partial_match and target_name in file_info['name']) or file_info['name'] == target_name:
            matching_paths.append(file_info['path'])
    
    return matching_paths

# Example usage
target_filename = "image_0.png"  # Example filename to search for
exact_matches = find_files_by_name(directory_structure, target_filename)
print(f"Found {len(exact_matches)} files with exact name '{target_filename}'")

# For partial matching (e.g., find all files with "image" in the name)
partial_matches = find_files_by_name(directory_structure, "sVal", partial_match=True)
print(f"Found {len(partial_matches)} files containing 'sVal' in their name")
```

```python
def group_images_by_focus(path_all_files):
    """
    Group images by focus and light from a list of file paths.
    
    Args:
        path_all_files (list): List of file paths
        
    Returns:
        dict: Dictionary mapping focus to list of (light, path) tuples
        list: Sorted list of focuses
    """
    from collections import defaultdict
    import os
    
    # Filter paths to only include sVal.png
    sval_paths = [path for path in path_all_files if 'sVal.png' in path]
    
    # Group images by focus
    focus_to_lights = defaultdict(list)
    
    for path in sval_paths:
        parts = path.split('/')
        
        # Find the light directory (format: LXXX)
        light_parts = [part for part in parts if part.startswith('L') and len(part) > 1 and part[1:].isdigit()]
        if not light_parts:
            continue
        light = light_parts[0]
        
        # Find the focus directory (format: zfXXX.XXXX-dfXXX.XXXX or sharp)
        focus_parts = [part for part in parts if part.startswith('zf')]
        if not focus_parts:
            continue
        focus = focus_parts[0]
        
        # Group paths by focus
        focus_to_lights[focus].append((light, path))
    
    # Sort each group by light
    for focus in focus_to_lights:
        focus_to_lights[focus].sort(key=lambda x: x[0])
    
    # Get sorted list of focuses for better organization
    sorted_focuses = sorted(focus_to_lights.keys())
    
    print(f"Found {len(sorted_focuses)} different focuses")
    for focus in sorted_focuses:
        print(f"  {focus}: {len(focus_to_lights[focus])} light variations")
    
    return focus_to_lights, sorted_focuses
```

```python
import os
import numpy as np
from PIL import Image


def average_images_by_focus(focus_to_lights, output_path):
    """
    Average images with the same focus but different light conditions.
    
    Args:
        focus_to_lights (dict): Dictionary mapping focus to list of (light, path) tuples
        output_path (str): Directory where averaged images will be saved
    """
    
    # Create output directory if it doesn't exist
    os.makedirs(output_path, exist_ok=True)
    
    # Track success and failure counts
    success_count = 0
    failure_count = 0
    
    # Process each focus level
    for focus, light_path_pairs in focus_to_lights.items():
        print(f"Processing {focus}...")
        
        # Extract image paths for this focus
        image_paths = [path for _, path in light_path_pairs]
        
        # Load all images for this focus
        images = []
        for path in image_paths:
            try:
                img = Image.open(path)
                images.append(img)
            except Exception as e:
                print(f"Error loading {path}: {e}")
        
        if not images:
            print(f"No valid images found for focus {focus}, skipping.")
            failure_count += 1
            continue
            
        try:
            # Convert images to numpy arrays
            arrays = [np.array(img) for img in images]
            
            # Check if all arrays have the same shape
            first_shape = arrays[0].shape
            if not all(arr.shape == first_shape for arr in arrays):
                print(f"Warning: Images for focus {focus} have different dimensions. Skipping.")
                failure_count += 1
                continue
                
            # Calculate average
            avg_array = np.mean(arrays, axis=0).astype(np.uint8)
            
            # Convert to PIL Image
            avg_image = Image.fromarray(avg_array)
            
            # Save the averaged image
            output_filename = f"{focus}.png"
            output_file_path = os.path.join(output_path, output_filename)
            avg_image.save(output_file_path)
            print(f"Saved averaged image for focus {focus} to {output_file_path}")
            success_count += 1
        except Exception as e:
            print(f"Error processing focus {focus}: {e}")
            failure_count += 1
    
    print(f"Finished processing all focus levels.")
    print(f"Successfully processed {success_count} focus levels, failed on {failure_count} focus levels.")
    
    return success_count, failure_count
```

```python
import os
import shutil

def organize_images_by_focus(focus_to_lights, output_dir):
    """
    Organize images by focus and rename them according to their light source.
    
    Args:
        focus_to_lights (dict): Dictionary mapping focus to list of (light, path) tuples
        output_dir (str): Path to output directory
    """
    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)
    
    # Track statistics
    copied_count = 0
    error_count = 0
    
    # Process each focus
    for focus, light_path_pairs in focus_to_lights.items():
        # Create a directory for this focus
        focus_dir = os.path.join(output_dir, focus)
        os.makedirs(focus_dir, exist_ok=True)
        
        # Process each light and path for this focus
        for light, path in light_path_pairs:
            # Define the destination filename (based on light)
            dest_filename = f"{light}.png"
            dest_path = os.path.join(focus_dir, dest_filename)
            
            # Copy the image
            try:
                shutil.copy2(path, dest_path)
                copied_count += 1
                print(f"Copied {path} to {dest_path}")
            except Exception as e:
                error_count += 1
                print(f"Error copying {path}: {e}")
    
    print(f"Images organized by focus in {output_dir}")
    print(f"Successfully copied {copied_count} images, encountered {error_count} errors")
```

```python
import os

def listar_todos_subdiretorios(diretorio):
    """
    Retorna uma lista com todos os subdiretórios em diferentes níveis de um diretório especificado.

    Args:
        diretorio (str): Caminho do diretório.

    Returns:
        list: Lista de todos os subdiretórios encontrados.
    """
    subdiretorios = []
    for root, dirs, _ in os.walk(diretorio):
        for nome in dirs:
            subdiretorios.append(os.path.join(root, nome))
    return subdiretorios

# Exemplo de uso:
# caminho = "/caminho/para/o/diretorio"
# subdirs = listar_todos_subdiretorios(caminho)
# print(subdirs)

# Exemplo de uso:
# caminho = "/caminho/para/o/diretorio"
# subdirs = listar_subdiretorios(caminho)
# print(subdirs)
```

```python
import os

def find_sval_images(directory):
    """
    Percorre a estrutura de diretórios e retorna uma lista com os caminhos completos
    de todos os arquivos chamados 'sVal.png'.

    :param directory: Caminho do diretório raiz a ser percorrido.
    :return: Lista de caminhos completos para os arquivos 'sVal.png'.
    """
    sval_images = []
    for root, _, files in os.walk(directory):
        for file in files:
            if file == "sVal.png" and "L000/zf" in root:
                sval_images.append(os.path.join(root, file))
    return sval_images

# Exemplo de uso:
directory_path = "/home/lelis/Documents/Projetos/hybrid-stereo-method/data/raw/hybrid_stereo/2025-03-08-stQ-melon24-amb0.00-glo0.50/stQ-melon14-amb0.00-glo0.50/0512x0384-hs01-kr10"
sval_images = find_sval_images(directory_path)
print(f"Encontrados {len(sval_images)} arquivos 'sVal.png':")
for path in sval_images:
    print(path)
```
