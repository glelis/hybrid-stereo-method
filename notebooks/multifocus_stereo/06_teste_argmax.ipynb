```python
def ler_csv_com_listas(caminho_arquivo):
    def converter_para_lista(valor):
        # Verifica se o valor é uma string que representa uma lista e converte
        try:
            return ast.literal_eval(valor) if valor.startswith('[') else valor
        except:
            return valor

    # Lê o arquivo CSV e converte as colunas com listas
    df = pd.read_csv(caminho_arquivo, converters={
        'x_list': converter_para_lista,
        'y_list': converter_para_lista,
        'w_list': converter_para_lista,
        'focus_values': converter_para_lista
    })

    return df


```

```python
def plot_multiple_quadratics(debug_info_df, indices):
    """
    Plota várias funções de segundo grau e seus respectivos pontos de dados a partir de um DataFrame.

    Args:
        debug_info_df: DataFrame contendo os dados.
        indices: Lista de índices das linhas do DataFrame a serem plotadas.
    """
    plt.figure(figsize=(10, 6))

    for idx in indices:
        row = debug_info_df.iloc[idx]
        A = row['A']
        B = row['B']
        C = row['C']
        x_list = row['x_list']
        y_list = row['y_list']

        x_values = np.linspace(min(x_list), max(x_list), 400)
        y_values = A * x_values**2 + B * x_values + C

        plt.plot(x_list, y_list, marker='o', linestyle='-', label=f'Data Points {idx}')
        plt.plot(x_values, y_values, linestyle='--', label=f'Quadratic Function {idx}')

    plt.xlabel('X values')
    plt.ylabel('Y values')
    plt.title('Plot of Multiple Quadratic Functions and Data Points')
    #plt.legend()
    plt.grid(True)
    plt.show()
```

```python
import os

def plot_all(debug_info_df, i, j, output_folder="output_plots"):
    """
    Plota todos os gráficos em uma única imagem, organizados em uma grade com i linhas e j colunas,
    e salva cada gráfico individualmente em uma pasta.

    Args:
        debug_info_df: DataFrame contendo os dados.
        i: Número de linhas na grade.
        j: Número de colunas na grade.
        output_folder: Pasta onde os gráficos serão salvos.
    """
    # Cria a pasta de saída, se não existir
    os.makedirs(output_folder, exist_ok=True)

    fig, axes = plt.subplots(i, j, figsize=(j * 5, i * 5))
    axes = axes.flatten()  # Para facilitar a iteração

    for idx, ax in enumerate(axes):
        if idx < len(debug_info_df):
            row = debug_info_df.iloc[idx]
            x_list = row['x_list']
            y_list = row['y_list']
            w_list = row['w_list']
            A = row['A']
            B = row['B']
            C = row['C']
            k_fuzzy = row['k_fuzzy']
            conf = row['conf']
            focus_values = row['focus_values']

            x_values = np.linspace(min(x_list), max(x_list), 400)
            y_values = A * x_values**2 + B * x_values + C

            x_max = -B/(2*A)
            y_max = A*x_max**2 + B*x_max + C

            #A, B, C = tuple(np.polyfit([i*0.1 for i in x_list], y_list, 2, w=w_list)) #coeficientes da funcao de segundo grau
            #print(-B/(2*A))

            ax.plot([i * 0.1 for i in range(len(focus_values))], focus_values, marker='o', linestyle='-', color='g', label='Focus Values')
            ax.plot([i * 0.1 for i in x_list], y_list, marker='o', linestyle='-', color='b', label='Data used (quadratic function)')
            ax.plot([i * 0.1 for i in x_values], y_values, linestyle='--', color='r', label='Quadratic Function')
            ax.set_yticks(np.arange(0, 1.1, 0.1))  # Define os intervalos do eixo y
            ax.set_xticks(np.arange(0, 1.1, 0.1))  # Define os intervalos do eixo x 

            
            # Adiciona os valores de f no gráfico
            for x, y in zip([i * 0.1 for i in range(len(focus_values))], focus_values):
                ax.text(x, y, f'{y:.5f}', fontsize=8, ha='right', color='black')

            ax.set_xlabel('Z')
            ax.set_ylabel('f')
            if row['pixel_j'] < 10:
                title = f"Z0{row['pixel_j']}-W0{row['pixel_i']}"
            else:
                title = f"Z{row['pixel_j']}-W0{row['pixel_i']}"
            ax.set_title(title)

            ax.legend(title=f"A: {A:.4f}, \nB: {B:.4f}, \nC: {C:.4f},\nw:{[round(i, 2) for i in w_list]} \nmax: {y_max:.4f}, \nk_fuzzy: {k_fuzzy*0.1:.4f}, \nconf: {conf:.4f}")
            ax.grid(True)

            # Salva o gráfico individualmente
            output_path = os.path.join(output_folder, f"{title}.png")
            fig_single, ax_single = plt.subplots(figsize=(8, 8))
            ax_single.plot([i * 0.1 for i in range(len(focus_values))], focus_values, marker='o', linestyle='-', color='g', label='Focus Values')
            ax_single.plot([i * 0.1 for i in x_list], y_list, marker='o', linestyle='-', color='b', label='Data used (Q.function)')
            ax_single.plot([i * 0.1 for i in x_values], y_values, linestyle='--', color='r', label='Quadratic Function')
            ax_single.set_yticks(np.arange(0, 1.1, 0.1))  # Define os intervalos do eixo y
            ax_single.set_xticks(np.arange(0, 1.1, 0.1))  # Define os intervalos do eixo x  


            # Adiciona os valores de f no gráfico
            for x, y in zip([i * 0.1 for i in range(len(focus_values))], focus_values):
                ax_single.text(x, y, f'{y:.5f}', fontsize=8, ha='right', color='black')

            ax_single.set_xlabel('Z')
            ax_single.set_ylabel('f')
            ax_single.set_title(title)
            ax_single.legend(title=f"A: {A:.4f}, \nB: {B:.4f}, \nC: {C:.4f},\nw:{[round(i, 2) for i in w_list]} \nmax: {y_max:.4f}, \nk_fuzzy: {k_fuzzy*0.1:.4f}, \nconf: {conf:.4f}")
            ax_single.grid(True)
            fig_single.savefig(output_path)
            plt.close(fig_single)
        else:
            ax.axis('off')  # Desativa o eixo para gráficos vazios

    plt.tight_layout()
    plt.show()
```
